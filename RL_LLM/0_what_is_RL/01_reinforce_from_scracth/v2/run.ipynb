```python
import os
import subprocess
import sys
from pathlib import Path

script_path = Path("train.py").resolve()

print(f"Running training script: {script_path}")

result = subprocess.run(
    [sys.executable, str(script_path)],
    cwd=str(script_path.parent),
    env=os.environ.copy(),
    capture_output=True,
    text=True,
)

if result.stdout:
    print(result.stdout)

if result.stderr:
    print("STDERR / ERRORS:")
    print(result.stderr)

if result.returncode != 0:
    raise subprocess.CalledProcessError(result.returncode, result.args)

print("Training finished successfully.")
```

This exposes something important about our toy environment:

Vanilla REINFORCE only reinforces trajectories according to their observed return.

Our environment gives:

good action → 1
bad action  → 0

so successful actions get reinforced while unsuccessful ones provide no gradient.

Later, with negative rewards or baselines, the behavior becomes more interesting.
