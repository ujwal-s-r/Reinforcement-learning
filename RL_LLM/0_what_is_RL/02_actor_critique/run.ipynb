Actor-Critic + Advantage

Now we upgrade our REINFORCE implementation.

The key change is:

REINFORCE
    ↓
return Gₜ
    ↓
policy update

becomes:

Actor-Critic

state
 ├────────→ Actor → action
 │
 └────────→ Critic → V(s)
                    ↓
                 advantage
                    ↓
                 Actor update

```python
import os
import subprocess
import sys
from pathlib import Path

script_path = Path("train_actor_critic.py").resolve()

print(f"Running training script: {script_path}")

result = subprocess.run(
    [sys.executable, str(script_path)],
    cwd=str(script_path.parent),
    env=os.environ.copy(),
    capture_output=True,
    text=True,
)

if result.stdout:
    print(result.stdout)

if result.stderr:
    print("STDERR / ERRORS:")
    print(result.stderr)

if result.returncode != 0:
    raise subprocess.CalledProcessError(result.returncode, result.args)

print("Training finished successfully.")
```
